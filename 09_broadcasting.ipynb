## 9. **Broadcasting**

Broadcasting is one of the most important NumPy concepts. It allows us to perform arithmetic and other operations on arrays with different shapes without manually resizing or repeating their elements.

### 1. **What is Broadcasting?**

Broadcasting is a NumPy mechanism that allows operations between arrays of different shapes when their dimensions are compatible.

NumPy automatically adjusts how arrays behave during an operation so that their shapes are compatible.

**Example:** Broadcasting with a scalar

````python
import numpy as np

arr = np.array([10, 20, 30, 40])

result = arr + 5

print(result)
````

**Understanding:**

NumPy treats the scalar 5 as if it were available for every element:
```
[10 20 30 40]
+  5  5  5  5
----------------
[15 25 35 45]
```

You do not need to manually create another array containing four **5**s.

### 2. **Broadcasting Rules**

NumPy compares array shapes from right to left (starting with the last dimension).

**Two dimensions are compatible when:**

1. Both dimensions have the same size.
2. One of the dimensions has size 1.

If neither condition is satisfied, broadcasting is not possible.

**Example 1:** Same dimensions

````python
arr1 = np.array([10, 20, 30])
arr2 = np.array([1, 2, 3])

print(arr1 + arr2)
print(arr1.shape)
print(arr2.shape)
````

Both arrays have shape **(3,)**, so they are compatible.

**Example 2:** One dimension has size **1**

````python
arr1 = np.array([[10], [20], [30]])
arr2 = np.array([1, 2, 3])

print(arr1.shape)
print(arr2.shape)

result = arr1 + arr2
print(result)
print(result.shape)
````

**Shapes:**

- arr1 → (3, 1)
- arr2 → (3,), treated as (1, 3) for comparison.

NumPy broadcasts them to a result with shape (3, 3).

**Example 3:** Incompatible shapes
```python
arr1 = np.array([10, 20, 30])
arr2 = np.array([1, 2])

print(arr1 + arr2)
```

**Output:**

`ValueError: operands could not be broadcast together with shapes (3,) (2,)`

The dimensions **3** and **2** are different, and neither is **1**. Therefore, broadcasting fails.

### 3. **Broadcasting Between Arrays**

Broadcasting allows NumPy to perform operations between arrays when their shapes are compatible.

**Example:** Adding a 1D array to a 2D array

````python
arr1 = np.array([
    [10, 20, 30],
    [40, 50, 60]
])

arr2 = np.array([1, 2, 3])

print(arr1.shape)
print(arr2.shape)

print(arr1 + arr2)
````

**Understanding:**

**Shapes:**
- arr1 → (2, 3)
- arr2 → (3,), treated as (1, 3)

### 4. **Broadcasting with Scalars**

A scalar is a single numerical value.

NumPy can perform operations between an array and a scalar. The scalar is automatically applied to every element.

**Example:**

````python
prices = np.array([100, 200, 300, 400])

print(prices + 50)
print(prices - 20)
print(prices * 2)
print(prices / 10)
````

### 5. **Broadcasting with Different Shapes**

Let's understand how broadcasting works with different array shapes.

**Example 1:** (3, 3) and (3,)

````python
arr1 = np.array([
    [10, 20, 30],
    [40, 50, 60],
    [70, 80, 90]
])

arr2 = np.array([1, 2, 3])

print(arr1 + arr2)
````

The 1D array is applied to every row.

**Example 2:** (3, 3) and (3, 1)

````python
arr1 = np.array([
    [10, 20, 30],
    [40, 50, 60],
    [70, 80, 90]
])

arr2 = np.array([
    [1],
    [2],
    [3]
])

print(arr1 + arr2)
````

The second array is applied to every column in its corresponding row.

### 6. **Practical Broadcasting Examples**

**Example 1:** Applying a discount to product prices

Suppose we have product prices and want to apply a 10% discount.

````python
prices = np.array([1000, 2000, 3000, 4000])

discount = 10

discounted_price = prices - (prices * discount / 100)

print(discounted_price)
````

The scalar discount percentage is applied to every product price.

**Example 2:** Adding monthly expenses to different categories

Suppose we have expenses for three categories across three months.

````python
expenses = np.array([
    [5000, 5500, 6000],
    [2000, 2500, 3000],
    [1000, 1200, 1500]
])

additional_expenses = np.array([500, 300, 200])

total_expenses = expenses + additional_expenses

print(total_expenses)
````
