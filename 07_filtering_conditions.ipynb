## 7. **Filtering and Conditional Operations**

Filtering and conditional operations help us select specific elements, check conditions, identify missing values, and clean NumPy arrays.

These operations are especially useful when working with real-world datasets.

Let's learn each operation with practical examples.

### 1. **Boolean Masks**
A Boolean mask is an array containing **`True`** and **`False`** values. It identifies which elements satisfy a condition.

**Example:**

````python
import numpy as np

marks = np.array([35, 45, 60, 75, 90])

mask = marks >= 60

print(mask)
````

Here:
- **`False`** means the condition is not satisfied.
- **`True`** means the condition is satisfied.

A Boolean mask can be used to filter the original array.

### 2. **Filtering Arrays**

Filtering means selecting only the elements that satisfy a condition.

**Example:**

````python
marks = np.array([35, 45, 60, 75, 90])

result = marks[marks >= 60]

print(result)
````

You can also filter using multiple conditions.

**Example:**

````python
marks = np.array([35, 45, 60, 75, 90])

result = marks[(marks >= 40) & (marks <= 75)]

print(result)
````

### 3. **`np.where()` – Conditional Selection**

**`np.where()`** selects values based on a condition.

It works in two common ways:

**A. Get the indexes where a condition is `True`**

````python
marks = np.array([35, 45, 60, 75, 90])

result = np.where(marks >= 60)

print(result)
````

The matching indexes are `2`, `3`, and `4`.

**B. Choose values based on a condition**

**Syntax:**
```python
np.where(condition, value_if_true, value_if_false)
```

**Example:**

````python
marks = np.array([35, 45, 60, 75, 90])

result = np.where(marks >= 60, 'Pass', 'Fail')

print(result)
````

Here:
- If marks are 60 or above → **`Pass`**.
- Otherwise → **`Fail`**.

### 4. **`np.select()` – Multiple Conditions**

**`np.select()`** allows us to apply multiple conditions and assign different results.

It is useful when we have more than two categories.

**Syntax:**
```python
np.select(conditions, choices, default=value)
```

**Example:**

````python
marks = np.array([35, 45, 60, 75, 90])

conditions = [
    marks >= 80,
    marks >= 60,
    marks >= 40
]

choices = [
    'Excellent',
    'Good',
    'Average'
]

result = np.select(conditions, choices, default='Fail')

print(result)
````

**Important:** When multiple conditions are **`True`**, **`np.select()`** uses the choice corresponding to the first matching condition.

### 5. **`np.any()` – Check If Any Condition Is `True`**

**`np.any()`** returns **`True`** if at least one element satisfies the condition.

**Example:**

````python
marks = np.array([35, 45, 60, 75, 90])

result = np.any(marks >= 80)

print(result)
````

At least one student scored 80 or above, so the result is **`True`**.

If no element satisfies the condition, it returns **`False`**.

### 6. **`np.all()` – Check If All Conditions Are `True`**

**`np.all()`** returns **`True`** only if every element satisfies the condition.

**Example:**

````python
marks = np.array([65, 75, 85, 90])

result = np.all(marks >= 40)

print(result)
````

All students scored 40 or above.

**Another example:**

````python
marks = np.array([35, 45, 60, 75])

print(np.all(marks >= 40))
````

One student scored below 40, so the result is **`False`**.

### 7. **`np.isin()` – Check Whether Values Exist**
**`np.isin()`** checks whether each element in an array exists in a given list of values.

It returns a Boolean array.

**Example:**

````python
arr = np.array([10, 20, 30, 40, 50])

result = np.isin(arr, [20, 40, 60])

print(result)
````

Here:
- 20 exists in the given list → `True`.
- 40 exists in the given list → `True`.
- Other values do not exist → `False`.

You can also use it to filter matching values.

````python
result = arr[np.isin(arr, [20, 40, 60])]

print(result)
````

**Real-world use:** Select orders belonging to specific order statuses or products belonging to selected categories.

### 8. **`np.unique()` – Find Unique Values**

**`np.unique()`** returns the distinct values in an array, removing duplicates.

**Example:**

````python
arr = np.array([10, 20, 10, 30, 20, 40, 30])

result = np.unique(arr)

print(result)
````

You can also count how many times each unique value appears using return_counts=**`True`**.

````python
arr = np.array([10, 20, 10, 30, 20, 40, 30])

values, counts = np.unique(arr, return_counts=True)

print(values)
print(counts)
````

### 9. **`np.isnan()` – Identify Missing Values**

**`np.isnan()`** checks whether array elements are **NaN** (Not a Number).

It returns **`True`** for missing numeric values represented by **`np.nan`**.

````python
arr = np.array([10, 20, np.nan, 40, np.nan])

result = np.isnan(arr)

print(result)
print(arr[np.isnan(arr)])
````

You can use it to identify and filter non-missing values.

````python
result = arr[~ np.isnan(arr)]
print(result)
````

**Real-world use:** Identify missing values in numerical datasets before performing calculations.

### 10. **`np.isinf()` – Identify Infinite Values**

**`np.isinf()`** checks whether array elements contain positive or negative infinity.

It returns **`True`** for both **`np.inf`** and **`-np.inf`**.

**Example:**

````python
arr = np.array([10, 20, np.inf, -np.inf, 50])

result = np.isinf(arr)

print(result)
````

````python
result = arr[~np.isinf(arr)]

print(result)
````

### 11. **`np.nan_to_num()` – Replace NaN and Infinite Values**

**`np.nan_to_num()`** replaces **NaN** and infinite values with numerical values.

**By default:**
- **NaN** is replaced with **`0`**.
- Positive infinity is replaced with the largest representable finite value.
- Negative infinity is replaced with the smallest representable finite value.

**Example:**

````python
arr = np.array([10, np.nan, 30, np.inf, -np.inf])

result = np.nan_to_num(arr)

print(result)
````

You can specify replacement values using **nan**, **posinf**, and **neginf**.

````python
arr = np.array([10, np.nan, 30, np.inf, -np.inf])

result = np.nan_to_num(
    arr,
    nan=0,
    posinf=100,
    neginf=-100
)

print(result)
````

This is particularly useful when cleaning numerical data before calculations.

### 12. **Practical Example – Sales Data Filtering**
Let's combine some of the operations we learned.

Suppose we have sales amounts and want to identify high-value sales, check missing values, and count unique sales amounts.

````python
sales = np.array([200, 500, 800, np.nan, 1200, 500, 200])

# Identify sales above 500
high_sales = sales[sales > 500]
print("High sales:", high_sales)

# Identify missing values
missing = np.isnan(sales)
print("Missing values:", missing)

# Replace missing values with 0
clean_sales = np.nan_to_num(sales, nan=0)
print("Clean sales:", clean_sales)

# Find unique sales amounts
unique_sales = np.unique(clean_sales)
print("Unique sales:", unique_sales)

# Check whether any sale is above 1000
print("Any sale above 1000:", np.any(clean_sales > 1000))

# Check whether all sales are non-negative
print("All sales non-negative:", np.all(clean_sales >= 0))
````
