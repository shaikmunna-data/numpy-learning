## 12. **Advanced NumPy Concepts**

These concepts help you understand how NumPy works internally, improve performance, and manage arrays efficiently.

### 1. **Vectorization**

Vectorization means performing operations on entire arrays instead of processing individual elements using Python loops.

NumPy performs these operations efficiently using optimized internal code.

**Example:**

Without vectorization (Python loop):

```python
numbers = [10, 20, 30, 40, 50]

result = []

for number in numbers:
    result.append(number * 2)

print(result)
```

With vectorization (NumPy):

```python
import numpy as np

numbers = np.array([10, 20, 30, 40, 50])

result = numbers * 2

print(result)
```

Key point: Vectorization allows you to perform operations on entire arrays without explicitly writing loops.

### 2. **Vectorized Operations**

Vectorized operations are operations that NumPy applies directly to array elements.

**Examples:**

```python
arr = np.array([10, 20, 30, 40, 50])

print(arr + 5)
print(arr * 2)
print(arr ** 2)
print(arr > 25)
```

### 4. **Views vs Copies**

This is an important NumPy concept.
- **View:** A new array that shares the same underlying data as the original array.
- **Copy:** A completely independent array with its own data.

**View example:**

```python
arr = np.array([10, 20, 30, 40, 50])

view_arr = arr[1:4]

view_arr[0] = 200

print(arr)
print(view_arr)
```

Changing the view also changes the original array because both share the same underlying data.

**Copy example:**

```python
arr = np.array([10, 20, 30, 40, 50])

view_arr = arr[1:4].copy()

view_arr[0] = 200

print(arr)
print(view_arr)
```

Changing the copy does not affect the original array.

**Key point:** **Views** share data; **copies** have independent data.

### 5. **Memory Management**

Memory management means understanding how NumPy arrays use and share computer memory.

**For example:**

```python
arr = np.array([10, 20, 30, 40, 50])

view_arr = arr[1:4]

copy_arr = arr[1:4].copy()

print(arr.nbytes)
print(view_arr.nbytes)
print(copy_arr.nbytes)
```

**Explanation:**

- Original array: 5 integers × 8 bytes = 40 bytes.
- View: 3 integers × 8 bytes = 24 bytes.
- Copy: 3 integers × 8 bytes = 24 bytes.

**Important:** `nbytes` reports the memory occupied by an array's elements. A view may report the same element memory size as its corresponding copy, even though the view shares the original data.

### 6. **`np.copy()`**

**`np.copy()`** creates an independent copy of an array.

**Example:**

```python
arr = np.array([10, 20, 30, 40, 50])

new_arr = np.copy(arr)

new_arr[0] = 100

print(arr)
print(new_arr)
```

You can also use:

```python
new_arr1 = arr.copy()

new_arr1[0] = 200

print(arr)
print(new_arr1)
```

Both approaches create a copy.

### 7. **`np.shares_memory()`**

**`np.shares_memory()`** checks whether two arrays share any underlying memory.

It returns a Boolean value:

- **`True`:** Arrays share memory.
- **`False`:** Arrays do not share memory.

**Example:**

```python
arr = np.array([10, 20, 30, 40, 50])

view_arr = arr[1:4]

copy_arr = arr[1:4].copy()

print(np.shares_memory(arr, view_arr))
print(np.shares_memory(arr, copy_arr))
```

**Key point:** Use this function to check whether two arrays share memory.

### 8. **Universal Functions (ufuncs)**

Universal functions (ufuncs) are NumPy functions that perform element-wise operations on arrays efficiently.

They work on individual elements without requiring explicit Python loops.

**Examples:**

```python
arr = np.array([1, 4, 9, 16, 25])

print(np.sqrt(arr))
print(np.square(arr))
print(np.exp(arr))
print(np.maximum(arr, 10))
```

**np.maximum()** compares each element of an array with another value and returns the larger value at each position.

### 9. **`np.vectorize()`**

**`np.vectorize()`** allows you to apply an ordinary Python function to each element of an array.

**Example:**

```python
def calculate_bonus(salary):
    if salary >= 40000:
        return salary * 0.10
    else:
        return salary * 0.05

salaries = np.array([20000, 30000, 40000, 50000])

vectorized_bonus = np.vectorize(calculate_bonus)

result = vectorized_bonus(salaries)

print(result)
```

However, for this particular calculation, NumPy already provides a better approach:

```python
result = np.where(salaries >= 40000,
                  salaries * 0.10,
                  salaries * 0.05)

print(result)
```

**Important:**  **`p.vectorize() provides`** convenient element-wise function application, but it is not true performance vectorization. It generally calls the Python function once per element and may not be faster than a Python loop.

For simple calculations, prefer normal NumPy operations:

```python
result = salaries * 0.10

print(result)
```
