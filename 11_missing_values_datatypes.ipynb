## 11. **Handling Missing Values and Data Types**

NumPy provides functions to identify and handle missing values and to manage the data types of array elements.

These concepts are useful when cleaning datasets and preparing data for analysis.
### 1. **`np.nan`**

**`np.nan`** means Not a Number. It is commonly used to represent missing or undefined numerical values in NumPy arrays.

**Example:**

````python
import numpy as np

arr = np.array([10, 20, np.nan, 40, 50])

print(arr)
````

Important points:
- `np.nan` represents a missing numerical value.
- It is a floating-point value.
- When a NumPy array contains both integers and np.nan, NumPy generally converts the array to a floating-point data type.
- `np.nan` is not equal to itself.

````python
print(np.nan == np.nan)
````

### 2. **`np.isnan()`**

**`np.isnan()`** checks whether array elements contain NaN values.

It returns a Boolean array.

**Example:**

````python
arr = np.array([10, 20, np.nan, 40, np.nan])

result = np.isnan(arr)

print(result)
````

Counting missing values:

````python
print(np.sum(np.isnan(arr)))
````

There are two missing values in the array.

### 3. **Handling Missing Values**

NumPy provides different ways to handle missing values.

One common approach is to filter out missing values using Boolean indexing.

**Example:** Removing missing values

````python
arr = np.array([10, 20, np.nan, 40, 50, np.nan])

cleaned_arr = arr[~ np.isnan(arr)]

print(cleaned_arr)
````

**Understanding:**

- **`np.isnan(arr)`** identifies missing values.
- **`~`** reverses the Boolean result.
- **`arr[~np.isnan(arr)]`** selects values that are not missing.

### 4. **Replacing Missing Values**

Sometimes we don't want to remove missing values. Instead, we want to replace them with another value.

We can use **`np.where()`**.

**Example:** Replacing NaN with zero

````python
arr = np.array([10, 20, np.nan, 40, 50, np.nan])

cleaned_arr = np.where(np.isnan(arr), 0, arr)

print(cleaned_arr)
````

**Understanding:**
```python
np.where(condition, value_if_true, value_if_false)
```
- If the condition is **`True`**, replace the value with **`0`**.
- Otherwise, keep the original value.

````python
arr = np.array([10, 20, np.nan, 40, 50, np.nan])

mean_value = np.nanmean(arr)

cleaned_arr = np.where(np.isnan(arr), mean_value, arr)

print(mean_value)
print(cleaned_arr)
````

**`np.nanmean()`** calculates the mean while ignoring missing values.

### 5. **Data Type Conversion**

Data type conversion means changing the data type of array elements.

**For example:**
- Integer to float.
- Float to integer.
- Integer to string.
- Integer to Boolean.

We can check an array's data type using **`.dtype`**.

**Example:**

````python
arr = np.array([10, 20, 30, 40])

print(arr.dtype)
````

The exact integer type may depend on your system.

### 6. **`astype()`**

**`astype()`** converts an array into a specified data type.

**Syntax:**

```python
array.astype(data_type)
```

````python
arr = np.array([10, 20, 30, 40])

float_arr = arr.astype(float)

print(arr.dtype, arr)

print(float_arr.dtype, float_arr)
````

**Example 2:** Float to integer

````python
arr = np.array([10.5, 20.8, 30.2, 40.9])

int_arr = arr.astype(int)

print(arr.dtype, arr)

print(int_arr.dtype, int_arr)
````

**Important:** Converting float to integer removes the decimal part. It does not round to the nearest integer.

**Example 3:** Integer to string

````python
arr = np.array([10, 20, 30])

str_arr = arr.astype(str)

print(arr.dtype, arr)

print(str_arr.dtype, str_arr)
````

The exact Unicode string dtype representation can vary depending on the values and NumPy version.

### 7. **Integer, Float, Boolean and String Arrays**

NumPy supports different data types for storing array elements.

**Example 1:** Integer array

````python
int_arr = np.array([10, 20, 30, 40])

print(int_arr)
print(int_arr.dtype)
````

**Example 2:** Float array

````python
float_arr = np.array([10.5, 20.5, 30.5])

print(float_arr)
print(float_arr.dtype)
````

**Example 3:** Boolean array

````python
bool_arr = np.array([True, False, True, True])

print(bool_arr)
print(bool_arr.dtype)
````

**Example 4:** String array

````python
str_arr = np.array(["Python", "NumPy", "Pandas"])

print(str_arr)
print(str_arr.dtype)
````

**`<U6`** means the array uses Unicode strings with a maximum stored length of 6 characters per element.

### 8. **Structured Arrays (Basic Understanding)**

A structured array allows us to store different data types in different fields of the same array.

Unlike a regular NumPy array, where elements generally share one data type, a structured array can contain fields such as:
- Employee ID – Integer
- Employee Name – String
- Salary – Float

**Example:**

````python
employees = np.array([
    (101, "Ravi", 30000.0),
    (102, "Priya", 40000.0),
    (103, "Arun", 35000.0)
], dtype=[
    ("employee_id", "i4"),
    ("employee_name", "U10"),
    ("salary", "f8")
])

print(employees)
````

**Accessing individual fields:**

````python
print(employees["employee_id"])
print(employees["employee_name"])
print(employees["salary"])
````
