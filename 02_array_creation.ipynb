## 2. **Creating NumPy Arrays**

NumPy provides different functions to create arrays. We can create arrays with zeros, ones, specific values, sequences, evenly spaced numbers and random numbers.

### 1. **np.array()**
`np.array()` creates a NumPy array from a Python list, tuple or another sequence.

**Syntax:**
```python
np.array(object, dtype=None)
```
**Example:**

````python
import numpy as np
numbers = np.array([10, 20, 30, 40, 50])
print(numbers)
````

### 2. **np.zeros()**
`np.zeros()` creates an array filled with zeros.

**Syntax:**
```python
np.zeros(shape, dtype=float)
```
**Example:**

````python
numbers = np.zeros(5)
print(numbers)
````

**2D Array:**

````python
numbers = np.zeros((3,4))
print(numbers)
````

**Specifying Data Type:**

````python
numbers = np.zeros(5, dtype=int)
print(numbers)
````

### 3. **np.ones()**
`np.ones()` creates an array filled with ones.

**Syntax:**
```python
np.ones(shape, dtype=float)
```

**Example:**

````python
numbers = np.ones(6)
print(numbers)
````

**2D Array:**

````python
numbers = np.ones((2,3), dtype=int)
print(numbers)
````

### 4. **np.empty()**

`np.empty()` creates an array without initializing its elements.

The array contains whatever values happen to be present in the allocated memory. These values are unpredictable and should not be treated as zeros.

**Syntax:**
```python
np.empty(shape, dtype=float)
```
**Example:**

````python
numbers = np.empty(5)
print(numbers)
````

The actual output varies each time because the memory is not initialized.

**2D Array:**

````python
numbers = np.empty((2, 3))

print(numbers)
````

**Output:**

The output contains unpredictable values in a 2 × 3 array.

- **Use:** When you want to create an array quickly and plan to fill every element before reading it.

### 5. **np.full()**

`np.full()` creates an array filled with a specific value.

**Syntax:**
```python
np.full(shape, fill_value, dtype=None)
```

**Example:**

````python
numbers = np.full(5, fill_value=10)
print(numbers)
````

**2D Array:**

````python
numbers = np.full((2,3), fill_value=7)
print(numbers)
````

### 6. **np.arange()**

`np.arange()` creates an array containing evenly spaced values within a given range.

The stop value is excluded.

**Syntax:**
```python
np.arange(start, stop, step)
```


**Parameters:**
- **start:** Starting value (default 0).
- **stop:** Ending value (excluded).
- **step:** Difference between consecutive values (default 1).

**Example:**

````python
numbers = np.arange(1,10)
print(numbers)
````

**Using Step:**

````python
even_numbers = np.arange(2,20,2)
print(even_numbers)
````

**2D Array using `np.arange()`**

`np.arange()` generates a sequence of numbers, and `reshape()` converts that sequence into a 2D array.

````python
numbers = np.arange(1, 11).reshape(2,5)
print(numbers)
````

**Using Float Values:**

````python
numbers = np.arange(0, 2, 0.5)

print(numbers)
````

### 7. **np.linspace()**

`np.linspace()` creates an array containing a specified number of evenly spaced values between a starting value and an ending value.

Unlike `np.arange()`, it specifies the number of values instead of the step size.

**Syntax:**
```python
np.linspace(start, stop, num=50)
```
**Parameters:**
- **start:** Starting value.
- **stop:** Ending value (included by default).
- **num:** Number of values to generate (default 50).

**Example:**

````python
numbers = np.linspace(1,10,5)
print(numbers)
````

````python
numbers = np.linspace(0, 100, 6)

print(numbers)
````

### 8. **np.eye()**

`np.eye()` creates a 2D array with ones on a specified diagonal and zeros elsewhere.

**Syntax:**
```python
np.eye(N, M=None, k=0, dtype=float)
```

**Parameters:**
- **N:** Number of rows.
- **M:** Number of columns (default N).
- **k:** Diagonal position (default 0).

**Example:**

````python
numbers = np.eye(3)

print(numbers)
````

**Another Example:**

````python
numbers = np.eye(5)

print(numbers)
````

**Rectangular Array:**

````python
numbers = np.eye(3, 4)

print(numbers)
````

**Using k:**

````python
numbers = np.eye(3, k=1)

print(numbers)
````

### 9. **np.identity()**

`np.identity()` creates a square identity matrix with ones on the main diagonal and zeros elsewhere.

**Syntax:**
```python
np.identity(n, dtype=float)
```
**Parameter:**

- **n:** Number of rows and columns.

**Example:**

````python
numbers = np.identity(3)

print(numbers)
````

**Specifying Data Type:**

````python
numbers = np.identity(4, dtype=int)

print(numbers)
````

### 10. **np.random.rand()**

`np.random.rand()` generates random numbers between 0 and 1 using a uniform distribution.

**Syntax:**
```python
np.random.rand(d0, d1, ...)
```

**Parameters:**

- **d0:** Size of the first dimension.
- **d1:** Size of the second dimension.

````python
numbers = np.random.rand(5)
print(numbers)
````

Actual values will be different each time.

**2D Array:**

````python
numbers = np.random.rand(2,3)
print(numbers)
````

### 11. **np.random.randint()**

`np.random.randint()` generates random integers within a specified range.

The high value is excluded.

**Syntax:**
```python
np.random.randint(low, high=None, size=None)
```
**Parameters:**
- **low:** Starting value (inclusive).
- **high:** Ending value (exclusive).
- **size:** Number or shape of random values.

**Example:**

````python
numbers = np.random.randint(1, 10, size=5)

print(numbers)
````

**2D Array:**

````python
numbers = np.random.randint(1, 100, size=(2, 3))

print(numbers)
````

Actual values will vary each time.

### 12. **np.random.randn()**

`np.random.randn()` generates random numbers from a standard normal distribution.

The distribution has:
- Mean = 0
- Standard deviation = 1

**Syntax:**
```python
np.random.randn(d0, d1, ...)
```
**Example:**

````python
numbers = np.random.randn(5)

print(numbers)
````

Values can be positive, negative or close to zero.

**2D Array:**

````python
numbers = np.random.randn(2, 3)

print(numbers)
````

Actual values will vary each time.

### 13. **np.random.default_rng()**

`np.random.default_rng()` creates a random number generator (RNG) that we can use to generate random numbers.

It is the recommended approach for generating random numbers in modern NumPy.

**Syntax:**
```python
rng = np.random.default_rng(seed=None)
```
**Parameter:**

- **seed:** Optional value used to generate reproducible random numbers.

**Example:**

````python
rng = np.random.default_rng()

numbers = rng.integers(1, 10, size=5)

print(numbers)
````

**Explanation:**

- **`np.random.default_rng()`** creates a random generator.
- **`rng`** stores that generator.
- **`rng.integers()`** generates random integers.
- **1** is the starting value (inclusive).
- **10** is the ending value (exclusive).
- **size=5** generates 5 numbers.

Every time you run the code, you will generally get different numbers.

**Using Seed:**

A seed produces reproducible random numbers.

````python
rng = np.random.default_rng(seed=42)

numbers = rng.integers(1, 10, size=5)

print(numbers)
````

Running the same code with the same seed produces the same output.

**2D arrays using default_rng()**

````python
rng = np.random.default_rng(42)

numbers = rng.integers(1, 100, size=(2, 3))

print(numbers)
````

### 14. **rng.random()**

`rng.random()` generates random floating-point numbers between 0 (inclusive) and 1 (exclusive) using NumPy's random number generator.

It is used with `np.random.default_rng()`.

````python
rng = np.random.default_rng()

numbers = rng.random(5)

print(numbers)
````
