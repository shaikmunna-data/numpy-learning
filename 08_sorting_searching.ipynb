### 8. **Sorting and Searching**

Sorting and searching functions help us arrange array elements, find the positions of minimum and maximum values, and search for specific values.

### 1. **`np.sort()`**

**`np.sort()`** returns a sorted copy of an array without changing the original array.

**Syntax:**
```python
np.sort(array, axis=-1)
```



**Example:**

````python
import numpy as np

arr = np.array([40, 10, 30, 20, 50])

sorted_arr = np.sort(arr)

print(sorted_arr)
print(arr)
````

Sorting in descending order:

````python
arr = np.array([40, 10, 30, 20, 50])

sorted_arr = np.sort(arr)[::-1]

print(sorted_arr)
````

**Sorting 2D arrays:**

By default, **`np.sort()`** sorts along the last axis (**`axis=-1`**).

````python
arr = np.array([
    [40, 80, 30],
    [60, 20, 50]
])

print(np.sort(arr, axis=1))
print(np.sort(arr, axis=0))
````

- **`axis=1`:** Sorts each row.
- **`axis=0`:** Sorts each column.

### 2. **`np.argsort()`**

**`np.argsort()`** returns the indices (positions) that would arrange an array in sorted order.

Unlike **`np.sort()`**, it returns positions instead of sorted values.

**Example:**

````python
arr = np.array([40, 10, 30, 20])

indices = np.argsort(arr)

print(indices)
print(arr[indices])
````

**Practical example:**

Suppose we have student marks and want to arrange student names according to their marks.

````python
names = np.array(["Ravi", "Priya", "Arun", "Neha"])
marks = np.array([75, 90, 60, 85])

indices = np.argsort(marks)

print(names[indices])
print(marks[indices])
````

### 3. **`np.argmax()`**

**`np.argmax()`** returns the index of the maximum value in an array.

**Example:**

````python
sales = np.array([1200, 2500, 1800, 3000, 2200])

index = np.argmax(sales)

print(index)
print(sales[index])
````

The maximum sales value is **3000**, and its index is **3**.

**Important:** **`np.argmax()`** returns the position, not the maximum value itself.

To get the maximum value, use **`np.max()`**.

### 4. **`np.argmin()`**

**`np.argmin()`** returns the index of the minimum value in an array.

**Example:**

````python
sales = np.array([1200, 2500, 1800, 3000, 2200])

index = np.argmin(sales)

print(index)
print(sales[index])
````

The minimum sales value is **1200**, and its index is **0**.

### 5. **`np.argmax()` with Axis**

For 2D arrays, **`np.argmax()`** can find the index of the maximum value along a particular axis.

**Example:**

````python
sales = np.array([
    [1200, 2500, 1800],
    [3000, 2200, 3500],
    [1500, 4000, 2800]
])

print(np.argmax(sales, axis=0))
print(np.argmax(sales, axis=1))
````

### 6. **`np.argwhere()`**

**`np.argwhere()`** returns the indices of array elements that satisfy a condition.

It returns the positions as a 2D array.

**Example:**

````python
arr = np.array([10, 20, 30, 40, 50])

indices = np.argwhere(arr > 25)

print(indices)
````

The values greater than **25** are **30**, **40**, **50**, located at indices **2**, **3**, **4**.

**Example with a 2D array:**

````python
arr = np.array([
    [10, 20, 30],
    [40, 50, 60]
])

indices = np.argwhere(arr > 30)

print(indices)
````

**Important**: **`np.argwhere()`** returns coordinates, whereas Boolean indexing returns the actual matching values.

### 7. **`np.searchsorted()`**

**`np.searchsorted()`** finds the index where a value should be inserted into a sorted array to maintain its sorted order.

**Syntax:**
```python
np.searchsorted(array, value, side="left")
```
**Example:**

````python
arr = np.array([10, 20, 30, 40, 50])

index = np.searchsorted(arr, 35)

print(index)
````

The value **35** should be inserted at index **3**.

Understanding **`side`**:

````python
arr= np.array([10, 20, 20, 30, 40])

print(np.searchsorted(arr, 20, side='left'))
print(np.searchsorted(arr, 20, side='right'))
````

- **`side="left"`:** Returns the first suitable insertion position.
- **`side="right"`:** Returns the position after existing equal values.

**Important:** The input array must already be sorted for the result to represent the correct insertion position.

### 8. **`np.unique()`**

**`np.unique()`** returns the unique elements from an array, in sorted order.

**Example:**

````python
arr = np.array([10, 20, 10, 30, 20, 40, 30])

unique_values = np.unique(arr)

print(unique_values)
````

**Getting unique values and their counts:**


Use **`return_counts=True`** to count how many times each unique value appears.

````python
arr = np.array([10, 20, 10, 30, 20, 40, 30])

unique_values = np.unique(arr, return_counts=True)

print(unique_values)

#Or

values, counts = np.unique(arr, return_counts=True)

print(values)
print(counts)
````

**Getting unique values and their first occurrence indices:**

````python
arr = np.array([10, 20, 10, 30, 20, 40, 30])

values, indices = np.unique(arr, return_index=True)

print(values)
print(indices)
````

**`return_index=True`** gives the original index of the first occurrence of each unique value.
