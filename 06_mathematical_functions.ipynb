## 6. **Mathematical and Statistical Functions**

NumPy provides mathematical and statistical functions to perform calculations on arrays.

These functions are useful in data analysis for calculating values, understanding data distributions, measuring relationships, and analyzing trends.

Let's learn each function with practical examples.

### 1. **`np.sqrt()` – Square Root**
Calculates the square root of each element in an array.

**Example:**

```python
import numpy as np

arr = np.array([4, 9, 16, 25, 36, 49])

result = np.sqrt(arr)

print(result)
```

For example:
- Square root of 4 = 2
- Square root of 9 = 3
- Square root of 16 = 4

### 2. **`np.abs()` – Absolute Value**

Returns the absolute (non-negative) value of each element.

It removes the negative sign from negative numbers.

**Example:**

```python
arr = np.array([-10, -5, 0, 5, 10])

print(np.abs(arr))
```

**Real-world use:** Calculate the difference between actual sales and target sales without considering whether the difference is positive or negative.

### 3. **`np.exp()` – Exponential**

Calculates the exponential value of each element.

It raises the mathematical constant _e_ (approximately 2.71828) to the power of each element.

**Example:**

```python
arr = np.array([0, 1, 2, 3])

print(np.exp(arr))
```

For example:
- `np.exp(0)` = 1
- `np.exp(1)` = 2.718
- `np.exp(2)` = 7.389

### 4. **`np.log()` – Logarithm**

Calculates the natural logarithm (base _e_) of each element.

**Example:**

```python
arr = np.array([1, 2.71828183,   7.3890561,   20.08553692])

print(np.log(arr))
```

### 5. **`np.round()` – Rounding**

Rounds numbers to the nearest value or to a specified number of decimal places.

**Example:**

```python
arr = np.array([10.456, 20.678, 30.123])

print(np.round(arr))
print(np.round(arr, 2))
```

The second argument specifies the number of decimal places.

### 6. **`np.floor()` – Floor**

Rounds each number down to the nearest integer (towards negative infinity).

**Example:**

```python
arr = np.array([10.8, 20.3, 30.9, -2.3])

print(np.floor(arr))
```

Notice that -2.3 becomes -3, not -2.

### 7. **`np.ceil()` – Ceiling**

Rounds each number up to the nearest integer (towards positive infinity).

**Example:**

```python
arr = np.array([10.2, 20.8, 30.1, -2.3])

print(np.ceil(arr))
```

### 8. **`np.cumsum()` – Cumulative Sum**

Calculates the running total of array elements.

Each output element represents the sum of all previous elements, including itself.

**Example:**

```python
arr = np.array([10, 20, 30, 40, 50])

print(np.cumsum(arr))
```

### 9. **`np.cumprod()` – Cumulative Product**

Calculates the running product of array elements.

**Example:**

```python
arr = np.array([2, 3, 4, 5])

print(np.cumprod(arr))
```

### 11. **`np.quantile()` – Quantiles**

Calculates the value below which a specified fraction of observations falls.

It is similar to **`np.percentile()`**, but uses fractions between 0 and 1 instead of percentages between 0 and 100.`

**Example:**

```python
marks = np.array([10, 20, 30, 40, 50])

print(np.quantile(marks, 0.25))
print(np.quantile(marks, 0.50))
print(np.quantile(marks, 0.75))
```

### 12. **`np.corrcoef()` – Correlation**

Measures the strength and direction of the linear relationship between two variables.

The result is a correlation matrix.

**Correlation values range from -1 to +1.**

<table style="margin-left: 0; margin-right: auto;">
<tr>
<th>Correlation</th>
<th>Meaning</th>
</tr>
<tr>
<td>+1</td>

</tr>
<tr>
<td>0</td>
<td>No linear correlation</td>
</tr>
<tr>
<td>-1</td>
<td>Perfect negative correlation</td>
</tr>

**Example:**

```python
study_hours = np.array([1, 2, 3, 4, 5])
marks = np.array([20, 30, 40, 50, 60])

result = np.corrcoef(study_hours, marks)

print(result)
```

**Understanding the output:**
- The first 1.0 represents the correlation of study hours with itself.
- The second 1.0 represents the correlation between study hours and marks.
- The third 1.0 represents the correlation between marks and study hours.
- The last 1.0 represents the correlation of marks with themselves.

### 13. **`np.cov()` – Covariance**
Measures how two variables change together.

- **Positive covariance:** Both variables tend to increase or decrease together.
- **Negative covariance:** One variable tends to increase when the other decreases.
- **Near-zero covariance:** Little linear co-movement.

**Example:**

```python
study_hours = np.array([1, 2, 3, 4, 5])
marks = np.array([20, 30, 40, 50, 60])

result = np.cov(study_hours, marks)

print(result)
```

**Understanding the output:**

- `2.5` = Variance of study hours.
- `250` = Variance of marks.
- `25` = Covariance between study hours and marks.

**Important:** **`np.cov()`** uses sample covariance by default (ddof=1). Its result depends on the units of the variables, unlike correlation, which is standardized between -1 and +1.
